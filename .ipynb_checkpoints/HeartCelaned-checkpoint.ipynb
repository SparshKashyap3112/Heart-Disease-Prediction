# Data Preprocessing and Cleaning

```python
df_cleaned = df.copy()
```

```python
df_cleaned.head()
```

## Convert categorical data into numbers

### Converting SEX in usable format

```python
df_cleaned['Sex'].value_counts()
```

```python
df_cleaned['Sex'] = df_cleaned['Sex'].map({"M" : 0, "F" : 1})
```

```python
df_cleaned.head()
```

### Converting ChestPainType in usable format

```python
df_cleaned['ChestPainType'].value_counts()
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns = ['ChestPainType'], drop_first=True)
```

```python
df_cleaned.head()
```

```python
bool_cols = ['ChestPainType_ATA', 'ChestPainType_NAP', 'ChestPainType_TA']

df_cleaned[bool_cols] = df_cleaned[bool_cols].fillna(False).astype(int)
```

```python
df_cleaned.head()
```

### Converting RestingECG in usable format

```python
df_cleaned['RestingECG'].value_counts()
```

```python
df_cleaned = pd.get_dummies(
    df_cleaned,
    columns=['RestingECG'],
    dtype=int
)
```

```python
df_cleaned[['RestingECG_LVH', 'RestingECG_Normal', 'RestingECG_ST']] = \
    df_cleaned[['RestingECG_LVH', 'RestingECG_Normal', 'RestingECG_ST']].astype(int)
```

```python
df_cleaned.head()
```

### Converting ST_Slope in usable format

```python
df_cleaned = pd.get_dummies(
    df_cleaned,
    columns=['ST_Slope'],
    dtype=bool
)
```

```python
st_cols = ['ST_Slope_Down', 'ST_Slope_Flat', 'ST_Slope_Up']

df_cleaned[st_cols] = df_cleaned[st_cols].astype(int)
```

```python
df_cleaned.head()
```

### Converting ST_Slope in usable format

```python
df_cleaned['ExerciseAngina'] = df_cleaned['ExerciseAngina'].map({
    'Y': 1,
    'N': 0
})
```

```python
df_cleaned.head()
```

## Standard Scaling

```python
from sklearn.preprocessing import StandardScaler

cols = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']

scaler = StandardScaler()

df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
```

```python
df_cleaned.head()
```

## FEATURE SELECTION

Feature selection = selecting the most useful existing features for your machine-learning model and removing unnecessary ones.

```python
from scipy.stats import pearsonr
import pandas as pd

selected_features = [
    'Age',
    'Sex',
    'RestingBP',
    'Cholesterol',
    'FastingBS',
    'MaxHR',
    'ExerciseAngina',
    'Oldpeak',
    'HeartDisease'
]

correlations = []

for feature in selected_features:
    if feature != 'HeartDisease':
        correlation, p_value = pearsonr(
            df_cleaned[feature],
            df_cleaned['HeartDisease']
        )

        correlations.append({
            'Feature': feature,
            'Pearson Correlation': correlation,
            'P-Value': p_value
        })

correlation_df = pd.DataFrame(correlations)

correlation_df = correlation_df.sort_values(
    by='Pearson Correlation',
    ascending=False
).reset_index(drop=True)

correlation_df
```

## Chi-Square Test for Feature Selection

The Chi-Square (χ²) test is used to check the relationship between categorical features and a categorical target.

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

cat_features = [
    'Sex',
    'FastingBS',
    'ExerciseAngina',
    'ChestPainType_ATA',
    'ChestPainType_NAP',
    'ChestPainType_TA',
    'RestingECG_LVH',
    'RestingECG_Normal',
    'RestingECG_ST',
    'ST_Slope_Down',
    'ST_Slope_Flat',
    'ST_Slope_Up'
]

chi2_results = {}

for col in cat_features:

    contingency = pd.crosstab(
        df_cleaned[col],
        df_cleaned['HeartDisease']
    )

    chi2_stat, p_val, dof, expected = chi2_contingency(contingency)

    decision = (
        "Reject Null (Keep Feature)"
        if p_val < alpha
        else "Fail to Reject Null (Drop Feature)"
    )

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_results_df = pd.DataFrame(chi2_results).T

chi2_results_df
```
